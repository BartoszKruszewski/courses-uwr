#### Zadanie 1 — LLM wygenerował Ci kod, sprawdź które fragmenty są poprawne, a które błędne — napraw błędne

Tak jak na zajęciach: poprosiliśmy model językowy o kod realizujący cele opisane przy każdym fragmencie. Część fragmentów jest poprawna, część — nie.
Błędy bywają głośne (wyjątek) i **ciche** (kod działa, ale wynik jest zły).

Dla każdego fragmentu:
1. uruchom go i zdecyduj: **OK** czy **BŁĄD**,
2. jeśli BŁĄD — napisz jednym zdaniem, na czym polega, i wpisz poprawioną wersję w komórce pod nim.

Każda komórka zaczyna się od wczytania danych od nowa, więc fragmenty nie wpływają na siebie nawzajem.

**Fragment 1.** Cel: Pierwsze 5 wierszy tabeli.

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)

# kod wygenerowany przez LLM
df.loc[0:5]
```

```python
# loc zwraca 6 wierszy (0..5).
df.head()
```

**Fragment 2.** Cel: Usuń z tabeli pomocniczą kolumnę `score_improvement`.

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)
df['score_improvement'] = df['exam_score'] - df['previous_scores']

# kod wygenerowany przez LLM
df = df.drop('score_improvement')
df.columns
```

```python
# drop domyślnie usuwa wiersze (axis=0), więc rzuca KeyError, trzeba wskazać kolumnę.
df = pd.read_csv(DATA_PATH)
df['score_improvement'] = df['exam_score'] - df['previous_scores']
df = df.drop(columns='score_improvement')
df.columns
```

**Fragment 3.** Cel: Wyniki poniżej 20 punktów zamień na 20 (pozostałe zostaw bez zmian).

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)

# kod wygenerowany przez LLM
df.loc[df['exam_score'] < 20, 'exam_score'] = 20
df['exam_score'].min()
```

**Fragment 4.** Cel: Studenci z frekwencją od 60% do 80% (włącznie).

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)

# kod wygenerowany przez LLM
wynik = df[(df['attendance_percent'] >= 60) | (df['attendance_percent'] <= 80)]
len(wynik), len(df)
```

```python
# użyto | zamiast & , więc warunek jest zawsze prawdziwy i zwraca wszystkich studentów.
wynik = df[(df['attendance_percent'] >= 60) & (df['attendance_percent'] <= 80)]
len(wynik), len(df)
```

**Fragment 5.** Cel: Zamień wpisane w ankiecie wyniki na liczby; nieczytelne wpisy mają stać się brakami (`NaN`).

```python
# wczytanie danych
ankieta = pd.Series(['85', '90', 'brak', '70', ' 65'])

# kod wygenerowany przez LLM
liczby = pd.to_numeric(ankieta)
liczby
```

```python
# to_numeric domyślnie rzuca ValueError na 'brak', a errors='coerce' zamienia nieczytelne wpisy na NaN.
liczby = pd.to_numeric(ankieta, errors='coerce')
liczby
```

**Fragment 6.** Cel: Wiersz studenta z najwyższym `exam_score`.

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)

# kod wygenerowany przez LLM
ranking = df.sort_values('exam_score', ascending=False)
ranking.loc[0]
```

```python
# po sortowaniu loc[0] zwraca wiersz o etykiecie 0 (S001), a nie pierwszy w rankingu, trzeba iloc[0].
ranking = df.sort_values('exam_score', ascending=False)
ranking.iloc[0]
```

**Fragment 7.** Cel: Studenci, którzy uczyli się ponad 10 godzin: tylko `student_id` i `exam_score`, rosnąco według wyniku.

```python
# wczytanie danych
df = pd.read_csv(DATA_PATH)

# kod wygenerowany przez LLM
df[df['hours_studied'] > 10][['student_id', 'exam_score']].sort_values('exam_score')
```
